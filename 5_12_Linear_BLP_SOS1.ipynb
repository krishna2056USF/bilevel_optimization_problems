```python
import gurobipy as gp
from gurobipy import GRB
```

```python
model = gp.Model("milp_sos1")

x = model.addVar(lb=0)
y = model.addVars(2, name="y")
lamda = model.addVars(5, lb=0, name="lamda")

g = model.addVars(5, lb=0, name="g")

model.addConstr(g[0] == 3 - 4*x - y[0] - y[1], "g1(x,y)")
model.addConstr(g[1] == 5 - 2*x + 2*y[0] - 5*y[1], "g2(x,y)")
model.addConstr(g[2] == 1 - 3*x + y[0] +  2*y[1], "g3(x,y)")
model.addConstr(g[3] == y[0], "g4_y1")
model.addConstr(g[4] == y[1], "g5_y1")

model.addConstr(lamda[0] - 2*lamda[1] - lamda[2] - lamda[3] == -2, "df1")
model.addConstr(lamda[0] + 5*lamda[1] - 2*lamda[2] - lamda[4] == 1, "df1")

for i in range(5):
    model.addSOS(GRB.SOS_TYPE1, [lamda[i], g[i]])

model.setObjective( 7*x - 2*y[0] + 3* y[1], GRB.MINIMIZE)

model.optimize()
```

```python
if model.Status == GRB.OPTIMAL:
    print(f"Objective = {model.ObjVal:.2f}")
    print(f"x = {x.X:.2f}")
    print(f"y1 = {y[0].X:.2f}")
    print(f"y2 = {y[1].X:.2f}")

```
