```python
import gurobipy as gp
from gurobipy import GRB
import numpy as np



from src.general_bilevel_branch_cut import (
    BranchAndCut,
    Knapsack915,
    PathInterdiction
)

# Proposition 9.7
solver97 = BranchAndCut(
    Knapsack915(),
    cut_strategy="prop97"
)
solver97.solve()
solver97.plot_tree("tree_prop97.png")

# Propositions 9.7 and 9.9 combined
solver_hybrid = BranchAndCut(
    Knapsack915(),
    cut_strategy="hybrid"
)
solver_hybrid.solve()

# Proposition 9.9 with linking-variable branching
solver99 = BranchAndCut(
    Knapsack915(),
    cut_strategy="prop99"
)
solver99.solve()
solver99.plot_tree("tree_prop99.png")
```
